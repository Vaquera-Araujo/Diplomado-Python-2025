# Clases, Constructores y Métodos

````python
# Ejemplo 1
class Particula(object):
  """
  Una partícula es la unidad más fundamental que constituye la materia

  Atributos
  ------------

  c: carga eléctrica en unidades de [e]
  m: masa en unidades de [kg]
  r: posición [m]
  """
  habla= "Hola, soy una partícula"
  #Constructor
  def __init__(self,carga,masa,posicion):
    """
    Inicializa la partícula con valores dados de carga, masa y posición
    """
    self.c=carga
    self.m=masa
    self.r=posicion
  # Métodos
  def propiedades(self):
    prop= self.habla+(
        ". Mi carga es: "+str(self.c)+
        ", mi masa es: "+str(self.m)+
        ", mi posición en x es: "+str(self.r['x'])+
        ", mi posición en y es: "+str(self.r['y'])+
        ", mi posición en z es: "+str(self.r['z']))

    print(prop)
````

````python
from scipy import constants
````

````python
proton=Particula(1,constants.m_p,{'x':1,'y':2,'z':-5})
````

````python
proton.propiedades()
````

````python
# Ejemplo 2
class Fruta(object):
  # Constructor
  def __init__(self,nombre,color):
    self.n=nombre
    self.c=color
  def presentar(self):
    print("la fruta es "+ self.n+" de color "+ self.c)
````

````python
fruta_1=Fruta("Manzana","Rojo")
````

````python
fruta_1.presentar()
````

# Herencia

En programación orientada a objetos se hace uso intensiv de la reutilización de código, de manera que podemos usar una clase preexistente para definir una clase nueva sin necesidad de crearla desde cero. La clase preexistente se denomina clase predecesora, mientras que la nueva clase se denomina subclase.

Sintaxis:
```
class clase_predecesora:
  <cuerpo de la clase predecesora>
class subclase(clase_predecesora)
  <cuerpo de la sublcase>  
```

````python
# Ejemplo
# Clase predecesora
class vehiculo(object):
  def info_vehiculo(self):
    print('pertenece a la clase vehículo')
# Subclase
class bicicleta(vehiculo):
  def info_bici(self):
    print('pertenece a la clase bicicleta')
````

````python
bici_1=bicicleta()
````

````python
bici_1.info_bici()
````

````python
bici_1.info_vehiculo()
````

# Polimorfismo

Existe más de una apariencia dentro de una misma especie. En la programación orientada a objetos, el polimorfismo ocurre cuando una clase hereda los atributos de una clase predecesora. Como regla general, lo que funciona para la clase predecesora debe funcionar también para la subclase, pero la subclase puede ejecutar su propio comportamiento especializado.

````python
# Ejemplo
# Clase predecesora
class Animal:
  def habla(self):
    pass
#Subclases
class Perro(Animal):
  def habla(self):
    return "Guau"
class Gato(Animal):
  def habla(self):
    return "Miau"
class Pato(Animal):
  def habla(self):
    return "Cuac"

# Función que hace hablar al animal y muestra el polimorfismo
def hacer_hablar(animal):
  print(animal.habla())

````

````python
can=Perro()
gat=Gato()
pat=Pato()
````

````python
hacer_hablar(can)
hacer_hablar(gat)
hacer_hablar(pat)
````

````python
cocodrilo=Animal()
````

````python
hacer_hablar(cocodrilo)
````
