Definición de funciones.

Sintaxis

```
def <nombre_funcion>(<argumento_1>,<argumento_2>,...):
   <código a ejecutar> # Obligatorio
   return <valor> # Opcional
```   

````python
# Función más simple, no hace absolutamente nada
def nada():
  pass
````

````python
nada()
````

````python
# Función Saludo
def saludo():
  print("¡Hola desde una función sin argumentos!")
````

````python
saludo()
````

````python
# Función con argumentos fijos
def suma(a,b):
  return a+b
````

````python
suma(4,5)
````

**Funciones con argurmentos variables**

Permiten flexibilidad con un número indefinido de argumentos no nombrados.

Sintaxis

```
def <nombre_funcion>(<argumento_1>,<argumento_2>,..., *<args>):
   <código a ejecutar> # Obligatorio
   return <valor> # Opcional
```   


````python
# Hay varias funciones precargadas que aceptan un número indefinido de variables
max(2,3,7,4,9,1)
````

````python
# Ejemplo 1
def lista_argumentos_1(*args):
  for i in args:
    print(i)
````

````python
lista_argumentos_1(1,2.0,"a")
````

````python
# Ejemplo 2
def lista_argumentos_2(*args):
  return args
````

````python
lista_argumentos_2(1,2.0,"a")
````

````python
type(lista_argumentos_2(1,2.0,"a"))
````

````python
# Ejemplo 3, Función mínimo artesanal.
def minimo(*args):
  m=args[0]
  for x in args[1:]:
    if x < m:
      m=x
  return m
````

````python
minimo(1,2,-1, 10, -4,-2)
````

Funciones con argumentos que toman valores predeterminados en caso de que el usuario no quiera cambiarlos (Argumentos claves o Keyword arguments).

Sintaxis

```
def <nombre_funcion>(<argumento_1>,<argumento_2>,...,<kwarg_1>=<val_1>,<kwarg_2>=<val_2>,...):
   <código a ejecutar> # Obligatorio
   return <valor> # Opcional
```   

````python
# Ecuación de una línea recta
def linea(x,m=1.0,b=0.0):
  return m*x+b
````

````python
linea(42) # No se modifican los argumentos claves, regresa 1.0*42+0.0
````

````python
linea(42,2) # No se modifica b, pero sí m, regresa 2*42+0.0
````

````python
linea(42,b=10) # No se modifica m, pero sí b, regresa 1.0*42+10
````

````python
linea(42,b=10, m=2) # cambiamos ambos argumentos claves, usando sus nombres
````

````python
linea(42,2,10) # cambiamos ambos argumentos claves siguiendo el orden en que se llaman en la función
````

También podemos declarar funciones que acepten un número indeterminado de argumentos clave.

Sintaxis

```
def <nombre_funcion>(<argumento_1>,<argumento_2>,...,**<kwargs>):
   <código a ejecutar> # Obligatorio
   return <valor> # Opcional
```   

````python
# Ejemplo 1
def lista_argumentos_clave_1(**kwargs):
  for llave,valor in kwargs.items():
    print(f"{llave}:{valor}")
````

````python
lista_argumentos_clave_1(nombre="Carlos",  ciudad="León", estado="Guanajuato" )
````

````python
# Ejemplo 2
def lista_argumentos_clave_2(**kwargs):
  return kwargs.items()
````

````python
lista_argumentos_clave_2(nombre="Carlos",  ciudad="León", estado="Guanajuato" )
````

Reto: Construir una función que calcule el promedio de los argumentos que se le otorguen

En Python solo es posible devolver un objeto, pero usando los contenedores adecuados, este objeto puede tenem múltiples valores.

````python
def momento_energia(m, v):
  p=m*v
  e=(1/2)*m*v**2
  return p, e
````

````python
# Esta función regresa una tupla
momento_energia(42.0,65.0)
````

````python
# Esta tupla se puede desempacar y su contenido se puede guardar en dos variables nuevas
mom, ener = momento_energia(42.0,65.0)
````

````python
mom
````

````python
ener
````

**Lambdas**

Funciones especiales diseñadas para crear operaciones en una sola línea sin usar la sentencia def. También se conocen como funciones anónimas y son muy útiles para simplificar el código y aumentar la legibilidad.

Sintaxis

```
lambda <argumentos>: <expresión>
```

````python
# Función que toma un argumento y calcula su cuadrado
(lambda x: x**2)(412)
````

````python
# Función que toma dos argumentos y calcula su suma
(lambda x,y: x+y)(5,3)
````

````python
# Podemos nombrar a estas funciones para invocarlas después
resta=lambda x,y: x-y
````

````python
resta(5,3)
````

````python
# Estas funciones admiten expresiones condicionales para realizar tareas más complejas
par_o_impar= lambda x: "Par" if x%2==0 else "Impar"
````

````python
par_o_impar(6)
````

````python
# También se peuden usar ciclos dentro de estas funciones anónimas
f= lambda:[x**2 for x in range(10)]
````

````python
f()
````
