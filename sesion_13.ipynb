**Arrays**
Son contenedores propios de Numpy, variantes de las listas pero con características propias:


*   Todos los elementos son del mismo tipo (preferentemente numérico entero, real o complejo), pues su propósito es el análisis numérico.
*   Tienen número de elementos determinado desde su creación.
*   Muchas operaciones pueden llevarse a cabo directamente en los arrays, simplificando el código y la evaluación numérica sin necesidad de implmentar bucles (vectorización).




````python
# Ejemplo 1
a1=np.array([6,28,496,8128])
a1
````

````python
type(a1)
````

````python
# Número de elementos
len(a1)
````

````python
a1[:2]
````

Funciones útiles para construir arrays son:


*   `array()`: Convierte una lista en array
*   `arange()`: Toma un valor inicial, uno final y un paso para crear una secuencia de números contenidos en un array.
*   `zeros()`: Genera un array lleno de ceros.
*   `ones()`: Genera un array lleno de unos.
*   `empty()`: Genera un array de cierto tamaño, asignando memoria a cada entrada, pero sin asignarle valores.





````python
# Ejemplos
np.arange(9)
````

````python
np.arange(3,99,3)
````

````python
np.zeros(10)
````

````python
np.ones(15)
````

````python
np.empty(6)
````

Operaciones con arrays

````python
# Ejemplos
x1=np.arange(0,100.25,0.25)
x1
````

````python
# Podemos implementar una función simultáneamente sobre todos los elementos del array
y1=(lambda x:x**3)(x1)
````

````python
y1
````

````python
# Podemos también usar las funciones definidas en numpy
z1=np.sin(x1)
z1
````

Otras funciones que inicializan arrays:


*   `linspace()`: crea arrays que representan una red de puntos separados linelmente.
*   `logspace()```: crea arrays con separación logarítmica.



````python
#Ejemplos
lin_1= np.linspace(1,2,10) # (valor inicial, valor final, número de elementos)
lin_1
````

````python
log_1=np.logspace(0,3,10) # (potencia inicial, potencia final, número de elementos)
log_1
````

````python
log_2=np.logspace(0,10,30,base=2) # se puede cambiar de base también
log_2
````

**Matplotlib**: Es el paquete estándar de Python para generar graficos.

Grafiquemos la curva $y=t² \exp(-t²)$, desde $t=0$ hasta $t=3$

````python
import matplotlib.pyplot as plt
````

````python
# Red de puntos para la variable t:

t_val=np.linspace(0,3,100)
````

````python
# Función
def y(t):
  return (t**2)*np.exp(-t**2)
````

````python
# Evaluación
y_val=y(t_val)
````

````python
y_val
````

````python
t_val
````

````python
# Gráfico y decoraciones
plt.plot(t_val,y_val,'b-') # (azul), 'r-' (rojo), 'g-' (verde)...  diferentes estilos de línea: o,,,.,+,x,_,s
plt.xlabel('t')
plt.ylabel('y')
plt.title('Matplotlib plot example')
plt.legend([r'$t^2\exp(-t^2)$'])
plt.show()
````
