# Función `reduce`

Aplicar una función de forma acumulativa a los elementos de una colección.

Sintaxis:` reduce(<función>,<collección>)`

````python
from functools import reduce
````

````python
#Ejemplo: implementar la suma de una lista de números
def sumar(x,y):
  return x+y
````

````python
numeros=[1,2,3,4,5]
````

````python
total=reduce(sumar, numeros)
print(total)
````

````python
# Mismo ejemplo usando una función lambda
total_lambda=reduce(lambda x,y:x+y, numeros)
print(total_lambda)
````

# Clases

Las clases definen colecciones lógicas de atributos que describen un tipo de objeto. También definen como crear un objeto particular de ese tipo. Su sintaxis es muy parecida a la de las funciones, se usa la sentencia `class`

```
class <nombre de la clase>(<objeto>):
  <cuerpo dela clase>
```

Una clase suele tener varios atibutos, entre los que se destacan:


*   Variables de clase: datos asociados a la clase misma
*   Constructores: Métodos especiales para inicializar un objeto que es una instancia de la clase
*   Métodos: Funciones especiales ligadas a un objeto específico que es una instancia de la clase

Las variables de instancia describen propiedades de un objeto específico dentro de una clase.



````python
# Ejemplo: clase que define objetos llamados partículas, con una variable de clase
class Particula:
  """Una partícula es la unidad más fundamental que constituye la materia"""
  habla="Hola, soy una partícula"
````

````python
help(Particula)
````

````python
Particula.habla
````

````python
proton=Particula()
````

````python
type(proton)
````

````python
proton.habla
````

````python
#Ejemplo de asignación de una variable de instancia a un objeto
proton.r={'x':0.01,'y':99.0,'z': 32.0}
````

````python
proton.r
````

# Constructores

Un constructor es una funcion que es ejecutada cuando un objeto de la clase es creado, y son útiles para que el objeto contenga desde su inicialización los atributos adecuados y no haya necesidad de implementarlos de manera manual.

````python
class Particula:
  """Una partícula es la unidad más fundamental que constituye la materia.

    Atributos
    ---------

    c: carga eléctrica en unidades de [e]
    m: masa en unidade de [kg]
    r: posición [m]
  """
  habla="Hola, soy una partícula"
  # Constuctor
  def __init__(self):
    """
    Inicializa la partícula con valores por default para sus atributos

    """
    self.c=0
    self.m=0.0
    self.r={'x':0.0,'y':0.0,'z': 0.0}
````

# Métodos

Los métodos son funciones ligadas a una definición de clase.

````python
class Particula(object):
  """Una partícula es la unidad más fundamental que constituye la materia.

    Atributos
    ---------

    c: carga eléctrica en unidades de [e]
    m: masa en unidade de [kg]
    r: posición [m]
  """
  habla="Hola, soy una partícula."
  # Constuctor
  def __init__(self):
    """
    Inicializa la partícula con valores por default para sus atributos

    """
    self.c=0
    self.m=0.0
    self.r={'x':0.0,'y':0.0,'z': 0.0}
  # Métodos
  def propiedades(self):
    prop= self.habla+(" Mi carga es: " +str(self.c)+ " Mi masa es: "+str(self.m)+
                      " Mi posición en x es: "+str(self.r['x'])+" Mi posición en y es: "+str(self.r['y'])+" Mi posición en z es: "+str(self.r['z']))
    print(prop)
````

````python
proton=Particula()
````

````python
proton.r
````

````python
proton.propiedades()
````
