# Sympy (continuación)

```python
from sympy import *
```

```python
x,y,z=symbols('x,y,z')
```

Funciones `factor`, `collect` y `logcombine:`

```python
#Ejemplo factor
expr_1= x**2-1
factor(expr_1)
```

```python
expr_1.factor()
```

```python
#Ejemplo collect
expr_2= x+x*z+x*y+y
collect(expr_2,x)
```

```python
expr_2.collect(x)
```

```python
#Ejemplo combine
a,b,c=symbols('a,b,c',positive=True)
expr_3=3*log(a)+log(b)-6*log(c)
```

```python
logcombine(expr_3)
```

**Sustituciones** con la función `subs`

```python
# Ejemplo 1
expr_6=x*z
expr_6
```

```python
expr_6.subs(x,y)
```

`subs` admite como argumento un diccionario completo, donde las llaves son remplazadas por sus valores

```python
# Ejemplo 2
expr_7=sin(expr_6)
expr_7
```

```python
expr_7.subs({sin:cos, x:y, z:exp(y)})
```

`subs` también sirve para sustituir valores numéricos en expresiones simbólicas

```python
# Ejemplo 3
expr_8=x*y+x*z**2-y*sqrt(z)+sin(z/x)
expr_8
```

```python
valores={x:1.5, y: 0.02, z:5.7}
```

```python
expr_8.subs(valores)
```

**Evaluación numérica**

Para evaluar numéricamente una expresión simbólica, podemos usar las funciones `N` o el método `evalf`. Ambas funciones toman un argumento opcional para indicar el número de cifras significativas en el resultado.

```python
# Ejemplo 1
N(pi/2)
```

```python
N(pi/2,100)
```

```python
# Ejemplo 2
(x+1/pi).evalf()
```

```python
(x+1/pi).evalf(5)
```

Otra manera de hacer evaluación numérica es a través de la función `lambdify`, que transforma una expresión simbólica completa en una función numérica lista para evaluarse.

```python
# Ejemplo 1
expr_9=sin(pi*x*exp(x))
expr_9
```

```python
expr_num=lambdify(x,expr_9)
```

```python
print(expr_num(0.3))
```

**Álgebra simbólica: Solución de ecuaciones y sistemas de ecuaciones**

Sympy puede resolver ecuaciones y sistemas de ecuaciones con la función `solve`

```python
# Ejemplo 1
solve(x**2+3*x+2)
```

```python
(x**2+3*x+2).subs(x,-2)
```

```python
(x**2+3*x+2).subs(x,-1)
```

```python
# Ejemplo 2
a,b,c=symbols('a,b,c',real=True)
solve(a*x**2+b*x+c,x)
```

```python
# Ejemplo 3
solve(sin(x)-cos(x),x)
```

```python
# Ejemplo 4
ec_1= x + 2*y -1
ec_2= x - y + 1
```

Para resolver un sistema, se introduce como primer argumento de `solve` una lista con las ecuaciones a resolver; como segundo argumento se inserta una lista con las variables, y se arroja como resultado un diccionario con la solución.

```python
# Ejemplo 5
solve([ec_1,ec_2],[x,y])
```

```python
# Ejemplo 6
ec_3= x**2 - y
ec_4= y**2 - x
sols=solve([ec_3,ec_4],[x,y],dict=True)
sols
```

```python
(ec_3.subs(sols[2])).simplify()
```
