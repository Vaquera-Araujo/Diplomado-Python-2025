Control de flujo en ciclos `for`

Palabras clave para controlar el flujo de ejecución de un ciclo:


*   `break` detiene el ciclo por completo
*   `continue` salta a la siguiente iteración
*   `pass` no hacer nada en la iteración presente
*   `else` ejecuta un código cuando la condición del ciclo deja de cumplirse




````python
frutas=["manzana", "pera", "durazno", "piña", "sandia", "banana", "uva", "fresa","naranja"]
````

````python
# Ejemplo brake
for f in frutas:
  if f=="sandia":
    break
  print(f)
print("El ciclo se ha roto aquí")
````

````python
# Ejemplo continue
for f in frutas:
  if f=="piña" or f=="naranja":
    continue
  print(f)
````

````python
# Ejemplo pass
for f in frutas:
  if f=="piña" or f=="naranja":
    pass
  print(f)
````

````python
contador=1
````

````python
# Ejemplo else
for f in frutas:
  print(contador, f)
  contador+=1
else: contador=1
````

**Ciclo While**

Permite ejecutar un bloque de código varias veces y es particularmente      útil cuando no tenemos una secuencia fija, sino una condición que puede ser verdadera o falsa. Este ciclo se ejecuta mientras la condición sea verdadera.

Sintaxis:

```
while <condición>:
  <código a ejecutar mientras la condición sea verdadera>
```

````python
while contador <=10:
  print(contador)
  contador+=1
else: contador=1
````

````python
#Ejemplo brake
contador=1
while contador <=10:
  if contador==5:
    break
  print(contador)
  contador+=1
````

````python
# Ejemplo continue
while contador<=10:
  if contador==5:
    contador+=1
    continue
  print(contador)
  contador+=1
else: contador=1
````

**Sentencia match-case**

Se utiliza cono una estructura de control que permite realizar comparaciones de patrones más sofisticadas y legibles en comparación con `if-elif-else`.

Sintaxis:
```
match <objeto>:
  case <patrón_1>:
    <código que se ejecuta cuando el objeto coincide con el patrón_1>
  case <patrón_2>:
    <código que se ejecuta cuando el objeto coincide con el patrón_2>  
  case _:
    <código que se ejecuta cuando el objeto no coincide con ningún patrón listado>



````python
#Ejemplo: manejo de menús
opcion=input("Elige una opción (1,2,3): ")
match opcion:
  case '1':
    print("Has elegido la opción 1.")
  case '2':
    print("Has elegido la opción 2.")
  case '3':
    print("Has elegido la opción 3.")
  case _:
    print("Opción inválida")
````

````python
# Ejemplo: clasificación de tipo de datos
dato=input("Inserta un dato: ")
match dato:
  case dato if dato.isdigit():
    print("Es un número")
  case dato if dato.isalpha():
    print("Es un texto")
````
