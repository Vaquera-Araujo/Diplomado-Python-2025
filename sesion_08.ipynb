Notas sobre definición de funciones

La funciones en python soy muy flexibles, podemos incluso combinar las sentencias `def` y `lambda`.

````python
# se puede definir un argumento clave con una lambda dentro de una función def
def func(vals, f=lambda x: sum(x)/len(x)):
  return f(vals)
````

````python
func([1,2,3,4,5,6])
````

Las funciones de Python son inherentemente recursivas, el nombre de una función puede usarse incluso dentro de la misma

````python
# Elementos de la secuencia de Fibonacci
def fib(n):
  if n==1 or n==2:
    return 1
  else:
    return fib(n-1)+fib(n-2)

````

````python
fib(25)
````

# Excepciones
En algunas ocasiones podemos anticiparnos a la evaluación de un error y evitar que el programa se dentenga cuando se encuentra uno. La excepciones nos permiten capturar y manejar errores de manera controlada.

Sintaxis

```
try:
  <código que puede generar una excepción>
except:
  <código que se ejecuta si se genera dicha excepción>
```  
dentro de esta estructura se pueden usar también las sentencias `else` para las situaciones donde no se genera la excepción, o `finally` para código que se ejecuta siempre, haya o no excepción.


Errores más comunes en Python:

1.   `ZeroDivisionError` División entre cero
2.   `ValueError` Cuando una función recibe un argumento con un valor inadecuado
3.   `TypeError` Cuando una función recibe un argumento con un tipo de dato inadecuado
4.   `IndexError` Cuando se intenta acceder a un índice fuera de rango
5.   `KeyError` Cuando se intenta accedar a una clave que no existe en un diccionario
6.   `FileNotFoundError` Cuando se intenta abrir un archivo que no existe



````python
# Ejemplo: Función que implementa una división segura
try:
  numerador=int(input("Introduce el numerador: "))
  denominador=int(input("Introduce el denominador: "))
  resultado= numerador/denominador
except ZeroDivisionError:
  print("No se puede dividir entre cero.")
else:
  print(f"{numerador} dividido entre {denominador} es {resultado}.")
finally:
  print("Operación completada")
````

````python
# Ejemplo: Manejo de datos en un formulario
try:
  edad= int(input("Ingrese su edad: "))
except ValueError:
  print("Este no es un número entero válido.")
else:
  print(f"Edad ingresada: {edad}")
finally:
  print("Operación de ingreso de edad completada")
````

# Función `map`

Aplica una función a cada elemento de una colección

Sintaxis: `map(<función>,<colección>)`

````python
# Ejemplo, elevar al cuadrado una colección de números
def cuadrado(x):
  return x**2

numeros=[1,2,3,4,5,6]
resultados=list(map(cuadrado,numeros))
````

````python
resultados
````

````python
# Mismo ejemplo usando una función lambda
resultados_lambda=list(map(lambda x:x**2,numeros))
````

````python
resultados_lambda
````

# Función `filter`

Filtra los elementos de una colección que no cumplen con una condición en la función de entrada.

Sintaxis: `filter(<función>,<colección>)`

````python
# Ejemplo: Seleccionar números pares dentro de una lista

def es_par(x):
  return x%2==0

pares=list(filter(es_par,numeros))
````

````python
pares
````

````python
# Mismo ejemplo con lambda

pares_lambda=list(filter(lambda x: x%2==0,numeros))
pares_lambda
````
